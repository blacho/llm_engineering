````python
# imports
import os
import requests
import json
from typing import List
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
from typing import Optional
````

````python
# constants
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
OLLAMA_API = "http://localhost:11434/api/chat"
BASE_URL = 'http://localhost:11434/v1'
HEADERS = {"Content-Type": "application/json"}
````

````python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if api_key and api_key.startswith('sk-proj-') and len(api_key)>10:
    print("API key looks good so far")
else:
    print("There might be a problem with your API key? Please visit the troubleshooting notebook!")
    
openai = OpenAI(base_url=BASE_URL)

code_system_prompt = (
    "You are senior code developer in  "
    "You are provided with a list of links found on a webpage."
    "You are able to decide which of the links would be most relevant to include in a brochure about the company"
    "such as links to an About page, or a Company page, or Careers/Jobs pages."
    "You should respond using markdown."
)
````

````python
def get_code_system_prompt(lang: str) -> str:
    return (
        f"You are the senior code developerin in the language {lang} that analyzes a given code and "
        "provids detailed explantion about ii in tersm of what it does and why.\n"
        "You can also suggest the code simplificaitons and enhancment for the code.\n"
        "You should respond in markdown."
    )
````

````python
# here is the question; type over this to ask something new
def get_user_question_promt(question: str) -> str:
    return (
        "Please explain what this code does and why:\n"
        f"{question}"
    )
````

````python
# Get gpt-4o-mini to answer, with streaming
# yield from {book.get("author") for book in books if book.get("author")}
````

````python
def get_code_prompt(lang: str, question: str) -> str:
    return [
        {'role': 'system', 'content': get_code_system_prompt(lang)},
        {'role': 'user', 'content': get_user_question_promt(question)},
    ]
````

````python
# Get Llama 3.2 to answer
# openai = OpenAI(base_url=base_url, api_key="ollama")

def code_assistent(lang: str, question: str, model: Optional[str] = None) -> None:
    if model:
        openai = OpenAI(base_url=BASE_URL, api_key="ollama")
    else:
        openai = OpenAI()
    stream = openai.chat.completions.create(
            model = model or MODEL_GPT,
            messages = get_code_prompt(lang, question),
            stream = True
        )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        response = response.replace("```","").replace("markdown", "")
        update_display(Markdown(response), display_id=display_handle.display_id)
````
